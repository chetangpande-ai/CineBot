```python
from typing import TypedDict
from langgraph.graph import StateGraph, START, END
```

```python
class State(TypedDict):
    name:str
    city:str
```

```python
def student_name_node(state:State):
    return {"name": f"{state['name']}"}

def student_city_node(state:State):
    return {"city": f"{state['city']}"}
```

```python
workflow=StateGraph(State)
```

```python
workflow.add_node("student_name", student_name_node)
workflow.add_node("student_city", student_city_node)
```

```python
workflow.add_edge(START,"student_name")
workflow.add_edge("student_name","student_city")
workflow.add_edge("student_city",END)
```

```python
app=workflow.compile()
```

```python
app.invoke({"name": "ishan", "city": "delhi"})
```

```python
from IPython.display import Image, display

display(Image(app.get_graph().draw_mermaid_png()))
```
